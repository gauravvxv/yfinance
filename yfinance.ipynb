# **Stock Price Monthly Analysis using Yfinance**

This project reads a list of stock tickers from an Excel file, downloads the last 5 years of daily stock price data using the `yfinance` API, resamples it to monthly data, and exports the results to an Excel file — one sheet per ticker.

Libraries used: `pandas`, `yfinance`, `openpyxl`

```python
import pandas as pd
import yfinance as yf
```

**Mount google drive**

## Load stock tickers from excel remove missing values and convert into python list

```python
input_path = '/content/drive/MyDrive/yfinance/tickers.xlsx'

ticker_df = pd.read_excel(input_path)
tickers = ticker_df['Tickers'].dropna().unique().tolist()
tickers
```

## Download 5 years of data and resample to monthly

```python
output_path = '/content/drive/MyDrive/yfinance/output.xlsx'

with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
    for ticker in tickers:
        print(f"Downloading data for {ticker}...")
        stock_data = yf.download(ticker, period='5y')

        if stock_data.empty:
            print(f"No data for {ticker}")
            continue

        stock_data['YearMonth'] = stock_data.index.to_period('M')
        stock_data.to_excel(writer,sheet_name = ticker[:31])

print("✅ Done! Monthly data saved in Excel.")
```

## ✅ Summary

- Loaded tickers from Excel
